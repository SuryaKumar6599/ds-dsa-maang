```python
"""
LLM Quantization Visualization

Visual comparison of memory savings and performance across quantization levels
"""

import json
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path


def main():
    # Load results
    results_file = Path("results/quantization_comparison.json")

    if results_file.exists():
        with open(results_file) as f:
            results = json.load(f)

        df = pd.DataFrame(results).T

        # Create subplots
        fig, axes = plt.subplots(2, 2, figsize=(12, 10))

        # Memory usage
        axes[0, 0].bar(df.index, df['memory_usage_gb'])
        axes[0, 0].set_ylabel('Memory (GB)')
        axes[0, 0].set_title('Memory Usage by Precision')
        axes[0, 0].tick_params(axis='x', rotation=45)

        # Speed comparison
        axes[0, 1].bar(df.index, df['tokens_per_second'])
        axes[0, 1].set_ylabel('Tokens/Second')
        axes[0, 1].set_title('Inference Speed')
        axes[0, 1].tick_params(axis='x', rotation=45)

        # Model size
        axes[1, 0].bar(df.index, df['model_size_gb'])
        axes[1, 0].set_ylabel('Model Size (GB)')
        axes[1, 0].set_title('Model Size on Disk')
        axes[1, 0].tick_params(axis='x', rotation=45)

        # Load time
        axes[1, 1].bar(df.index, df['load_time_s'])
        axes[1, 1].set_ylabel('Load Time (s)')
        axes[1, 1].set_title('Model Load Time')
        axes[1, 1].tick_params(axis='x', rotation=45)

        plt.tight_layout()

        # Save figure
        output_path = Path("results/quantization_comparison.png")
        output_path.parent.mkdir(parents=True, exist_ok=True)
        plt.savefig(output_path, dpi=300, bbox_inches='tight')

        plt.show()

        print(f"📊 Visualizations saved to {output_path}")

    else:
        print("❌ No results found. Run compare_quantization.py first.")


if __name__ == "__main__":
    main()
```
